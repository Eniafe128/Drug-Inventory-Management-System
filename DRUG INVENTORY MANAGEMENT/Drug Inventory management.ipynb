# Drug Inventory Management System 

## Project Tasks 
## Task 1 — Create the Variables 

```python
drug_name = "Paracetamol" 
price = 850
quantity_in_stock = 250 
quantity_sold_today = 65 
minimum_stock = 100 
expired = False

print(drug_name)
print(price)
print(quantity_in_stock)
print(quantity_sold_today)
print(minimum_stock)
print(expired)
```

## Task 2 — Calculate Remaining Stock 

```python
remaining_stock = quantity_in_stock - quantity_sold_today
print(remaining_stock)
```

## Task 3 — Calculate Sales Revenue

```python
revenue = price * quantity_sold_today
print(revenue)
```

## Task 4 — Update Inventory after today

```python
updated_stock = remaining_stock
print(updated_stock)
```

## Task 5 — Restocking Check

```python
reorder_drug = updated_stock < minimum_stock
print(reorder_drug)
```

## Task 6 — Expiry Check 

```python
expired = False
print(expired)
```

## Task 7 — Final Approval

```python
selling_criteria = (updated_stock > minimum_stock) and expired == False
print("Pharmacist can continue selling the drug:", selling_criteria)
```

## Task 8 — Discount Eligibility

```python
stock = (quantity_in_stock > 200) or (revenue > 50000)
print(stock)
```

## Task 9 — Inventory Summary

```python
inventory_report = """
========== DRUG INVENTORY REPORT ==========
"""
print(inventory_report)
print("Drug Name:", drug_name)
print("Price per unit:", price)
print("Current Stock:", quantity_in_stock)
print("Units Sold Today:", quantity_sold_today)
print("Today's Revenue:", revenue)
print("Needs Restocking:", reorder_drug)
print("Expired:", expired)
print("Safe to Sell:", selling_criteria)
print(stock)
print("============================================")
```
