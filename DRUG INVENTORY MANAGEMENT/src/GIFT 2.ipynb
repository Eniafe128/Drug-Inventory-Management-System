#  Hospital Appointment Management System

```python
# TASK 1 — Create Variables
# Each variable stores one piece of information about the appointment.
patient_name = "Sarah Johnson"
patient_id = "PT001"
patient_age = 42
department = "Cardiology"
doctor_name = "Dr. Musa"
appointment_day = "Monday"
appointment_time = "10:00 AM"
emergency_status = False


# TASK 2 — Validate Patient Information

name_valid = patient_name != ""
patient_id_valid = patient_id != ""
# != means not equal to, it asks if the patient's name and ID NOT empty of which is true
age_valid = patient_age > 0 and patient_age <= 120

print("Name valid:", name_valid)
print("Patient ID valid:", patient_id_valid)
print("Age valid:", age_valid)


# TASK 3 — Validate Department

if department == "Cardiology":
    print("Department Available")

elif department == "Pediatrics":
    print("Department Available")

elif department == "General Medicine":
    print("Department Available")

elif department == "Orthopedics":
    print("Department Available")

elif department == "Neurology":
    print("Department Available")

else:
    print("Invalid Department")
    
# Python checks the conditions from top to bottom. Since Sarah's department is cardiology, there's no need to check the remaining elif statements.


# TASK 4 — To Categorize Patients 
if patient_age <= 12:
    category = "Child"

elif patient_age <= 17:
    category = "Teenager"

elif patient_age <= 64:
    category = "Adult"

else:
    category = "Senior"

print("Category:", category)


# TASK 5 — Determine Appointment Priority

if emergency_status == True:
    priority = "High Priority"

elif category == "Senior":
    priority = "Medium Priority"

else:
    priority = "Normal Priority"
# Sarah's emergency_status = False and category = "Adult", emergency_status is false and category as senior is false. 
#Therefore, python reaches else and assign Normal priority
print("Priority Level:", priority)


# TASK 6 — Create Validation Functions
# 1. validate_age()
def validate_age(age):
    if age > 0 and age <= 120:
        return True
    else:
        return False
        
# 2. validate_department()      
def validate_department(department):
    if department == "Cardiology":
        return True
    elif department == "Pediatrics":
        return True
    elif department == "General Medicine":
        return True
    elif department == "Orthopedics":
        return True
    elif department == "Neurology":
        return True
    else:
        return False
# The function receives the department as a parameter and checks it against the hospital's allowed department

# 3. validate_patient_id()
def validate_patient_id(patient_id):
    if patient_id != "":
        return True
    else:
        return False
# A patient ID is valid only if it isn't empty
```

```python
# TASK 5 — Create an Appointment Function

def create_appointment(name, patient_id, age, department,
                       doctor_name, appointment_day, appointment_time):

    # Validate patient information
    name_valid = name != ""
    patient_id_valid = validate_patient_id(patient_id)
    age_valid = validate_age(age)
    department_valid = validate_department(department)
    
    # Determine patient category
    if age <= 12:
        category = "Child"
    elif age <= 17:
        category = "Teenager"
    elif age <= 64:
        category = "Adult"
    else:
        category = "Senior"

    # Determine priority
    emergency_status = False
    
    if emergency_status == True:
        priority = "High Priority"
    elif category == "Senior":
        priority = "Medium Priority"
    else:
        priority = "Normal Priority"
    
    # Determine appointment status
    if name_valid and patient_id_valid and  age_valid and department_valid:
        status = "Confirmed"
    else:
        status = "Invalid"

    return name, patient_id, age, category, department, doctor_name, \
           appointment_day, appointment_time, priority, status
# Instead of writing the same code repeatedly for every patient, we use function for reusability by calling create_appointment
```

```python
# TASK 8 — Use a for Loop

# The four patients provided by the project are given below,therefore John Smith's appointment should be invalid since its not part of the department.
patients = [
    ["John Smith", 45, "Surgery"],
    ["James Bello", 70, "Neurology"],
    ["Ada Obi", 8, "Pediatrics"],
    ["Sarah Johnson", 42, "Cardiology"]
]

for patient in patients:

    name = patient[0]
    age = patient[1]
    department = patient[2]

    name_valid = name != ""
    patient_id_valid = True
    # Validate age
    age_valid = validate_age(age)

    # Validate department
    department_valid = validate_department(department)

    # Determine category
    if age <= 12:
        category = "Child"
    elif age <= 17:
        category = "Teenager"
    elif age <= 64:
        category = "Adult"
    else:
        category = "Senior"

    # Determine priority
    if emergency_status == True:
        priority = "High Priority"
    elif category == "Senior":
        priority = "Medium Priority"
    else:
        priority = "Normal Priority"

    # Determine appointment status
    if name_valid and patient_id_valid and age_valid and department_valid:
        status = "Confirmed"
    else:
        status = "Invalid"

    print("----------------------------")
    print("Patient:", name)
    print("Age:", age)
    print("Category:", category)
    print("Department:", department)
    print("Priority:", priority)
    print("Status:", status)

# John's appointment is invalid because Surgery is not among the five permitted departments
```

```python
# TASK 9 — Use a While Loop 

patient_id = ""

while patient_id == "":
    patient_id = input("Enter Patient ID: ")

    if patient_id == "":
        print("Patient ID cannot be empty.")

print("Patient ID accepted:", patient_id)
```

```python
# TASK 10 — Infinite While Loop

while True:

    answer = input("Do you want to register another appointment? ")

    if answer == "Yes":
        print("Continue registration.")

    elif answer == "No":
        print("Registration stopped.")
        break

    else:
        print("Please enter Yes or No.")
```
